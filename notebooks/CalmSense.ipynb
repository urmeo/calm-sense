# CalmSense: synthetic calibration demo

Run feature extraction, leave-one-subject-out evaluation, and probability calibration on synthetic WESAD-shaped signals. Synthetic results check the pipeline; they do not reproduce the research benchmark.


## 1. Install

````python
import os
import sys
import subprocess
from pathlib import Path

repo = Path.cwd()
if not (repo / 'scripts/calibration.py').is_file():
    repo = repo / 'calm-sense'
    if not repo.exists():
        subprocess.run(['git', 'clone', '-q', 'https://github.com/urmeo/calm-sense.git', str(repo)], check=True)
    if not (repo / 'scripts/calibration.py').is_file():
        raise RuntimeError(f'{repo} is not a CalmSense checkout')
os.chdir(repo)
subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', '-r', 'requirements.txt'], check=True)
````

## 2. Run calibration on synthetic data

Generate synthetic signals, extract features, and compare probabilities before and after recalibration. No dataset download is needed.

````python
subprocess.run([sys.executable, 'scripts/calibration.py', '--synthetic'], check=True)
````

````python
from IPython.display import Image, display
for fig in ['calibration_reliability', 'calibration_gap', 'calibration_decision_curve']:
    display(Image(f'outputs/generated/demo/figures/{fig}.png'))
````

## 3. Run on the real datasets

Review the WESAD research agreement before downloading. The archive is about 2 GB; the recorded subject pickles occupy about 13.8 GB after extraction. Allow additional space for the archive and generated files. PhysioNet Non-EEG is only needed for transfer analysis.

```bash
python scripts/download_data.py --wesad
python scripts/download_data.py --verify-wesad
python scripts/run_experiment.py

# Optional: data for cross-dataset transfer
python scripts/download_data.py --noneeg
python scripts/download_data.py --verify-noneeg
```

You can also place the official WESAD files under `data/raw/WESAD/` before verification. Real-data experiments replace saved results and models. The [README run instructions](https://github.com/urmeo/calm-sense#run) cover setup and the benchmark; the [evaluation table](https://github.com/urmeo/calm-sense#evaluation) summarizes the recorded evaluation.
